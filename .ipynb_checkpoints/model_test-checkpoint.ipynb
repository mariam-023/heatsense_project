```python
START_DATE = "2013-01-01"
END_DATE = "2026-09-23"
```

```python
def mask_landsat(image):
    qa = image.select("QA_PIXEL")

    cloud = qa.bitwiseAnd(1 << 3).eq(0)
    cloud_shadow = qa.bitwiseAnd(1 << 4).eq(0)

    return image.updateMask(cloud).updateMask(cloud_shadow)
```

```python
def process_landsat(image):

    image = mask_landsat(image)

    # Surface temperature
    lst = (
        image.select("ST_B10")
        .multiply(0.00341802)
        .add(149.0)
        .subtract(273.15)
        .rename("LST")
    )

    # NDVI
    ndvi = (
        image.normalizedDifference(["SR_B5", "SR_B4"])
        .rename("NDVI")
    )

    # NDBI
    ndbi = (
        image.normalizedDifference(["SR_B6", "SR_B5"])
        .rename("NDBI")
    )

    return image.addBands([lst, ndvi, ndbi])
```

```python
l8 = (
    landsat8
    .filterBounds(karnataka)
    .filterDate(START_DATE, END_DATE)
    .map(process_landsat)
)

l9 = (
    landsat9
    .filterBounds(karnataka)
    .filterDate(START_DATE, END_DATE)
    .map(process_landsat)
)

landsat = l8.merge(l9)

print("Number of Landsat images:")
print(landsat.size().getInfo())
```

```python
landsat_composite = (
    landsat
    .select(["LST", "NDVI", "NDBI"])
    .median()
    .clip(karnataka)
)

print("Landsat composite created.")
```

```python
era = (
    era5
    .filterBounds(karnataka)
    .filterDate(START_DATE, END_DATE)
)

print("ERA5-Land images:")
print(era.size().getInfo())
```

```python
def process_era5(image):

    air_temp = (
        image.select("temperature_2m")
        .subtract(273.15)
        .rename("Air_Temperature")
    )

    dew_temp = (
        image.select("dewpoint_temperature_2m")
        .subtract(273.15)
    )

    u = image.select("u_component_of_wind_10m")
    v = image.select("v_component_of_wind_10m")

    wind = (
        u.pow(2)
        .add(v.pow(2))
        .sqrt()
        .rename("Wind_Speed")
    )

    # Magnus formula
    rh = (
        dew_temp.expression(
            """
            100 * exp((17.625 * Td) / (243.04 + Td) -
                      (17.625 * T) / (243.04 + T))
            """,
            {
                "Td": dew_temp,
                "T": air_temp
            }
        )
        .rename("Relative_Humidity")
    )

    return image.addBands([
        air_temp,
        rh,
        wind
    ])
```

```python
era_processed = era.map(process_era5)

era_composite = (
    era_processed
    .select([
        "Air_Temperature",
        "Relative_Humidity",
        "Wind_Speed"
    ])
    .mean()
    .clip(karnataka)
)

print("ERA5 environmental variables calculated.")
```

```python
def normalize_image(image, region, scale=1000):
    band_name = ee.String(image.bandNames().get(0))

    stats = image.reduceRegion(
        reducer=ee.Reducer.minMax(),
        geometry=region,
        scale=scale,
        maxPixels=1e9,
        bestEffort=True
    )

    min_value = ee.Number(stats.get(band_name.cat("_min")))
    max_value = ee.Number(stats.get(band_name.cat("_max")))

    normalized = (
        image
        .subtract(min_value)
        .divide(max_value.subtract(min_value))
        .clamp(0, 1)
    )

    return normalized
```

```python
region = karnataka.geometry()

lst_n = normalize_image(raw_lst, region).rename("LST_n")

air_n = normalize_image(raw_air, region).rename("AirT_n")

ndbi_n = normalize_image(raw_ndbi, region).rename("NDBI_n")

rh_n = normalize_image(raw_rh, region).rename("RH_n")

ndvi_n = normalize_image(raw_ndvi, region).rename("NDVI_n")

wind_n = normalize_image(raw_wind, region).rename("Wind_n")
```

```python
worldcover = (
    ee.Image("ESA/WorldCover/v200/2021")
    .clip(karnataka)
)

lulc_heat = worldcover.remap(
    [10, 20, 30, 40, 50, 60, 70, 80, 90, 95, 100],
    [0.1, 0.2, 0.3, 0.5, 1.0, 0.8, 0.4, 0.0, 0.2, 0.1, 0.3]
).rename("LULC_Heat")

print("LULC heat created:")
print(lulc_heat.bandNames().getInfo())
```

```python
chi = (
    lst_n.multiply(0.25)
    .add(air_n.multiply(0.20))
    .add(ndbi_n.multiply(0.15))
    .add(lulc_heat.multiply(0.15))
    .add(rh_n.multiply(0.10))
    .add(ndvi_n.multiply(-0.10))
    .add(wind_n.multiply(-0.05))
    .add(0.15)
    .rename("CHI")
)

print("CHI created successfully!")
```

```python
chi = (
    lst_n.multiply(0.25)
    .add(air_n.multiply(0.20))
    .add(ndbi_n.multiply(0.15))
    .add(lulc_heat.multiply(0.15))
    .add(rh_n.multiply(0.10))
    .add(ndvi_n.multiply(-0.10))
    .add(wind_n.multiply(-0.05))
    .add(0.15)
    .rename("CHI")
)
```

```python
training_image = (
    lst_n
    .addBands(ndvi_n)
    .addBands(ndbi_n)
    .addBands(air_n)
    .addBands(rh_n)
    .addBands(wind_n)
    .addBands(lulc_heat)
    .addBands(chi)
)

print(training_image.bandNames().getInfo())
```

```python
samples = training_image.sample(
    region=karnataka.geometry(),
    scale=1000,
    numPixels=500,
    seed=42,
    geometries=False
)

data = samples.getInfo()

rows = [
    feature["properties"]
    for feature in data["features"]
]

df = pd.DataFrame(rows)

print("Number of samples:", len(df))
df.head()
```
